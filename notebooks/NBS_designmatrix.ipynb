```python
# how to remove first two columns of file
# Define the input and output file paths
input_file_path =  '../../NBS/Design_Matrices/DesignMatrixHCP.txt'
output_file_path = '../../NBS/Design_Matrices/DesignMatrixHCPAgeSex.txt'

# Open the input file for reading and the output file for writing
with open(input_file_path, 'r') as infile, open(output_file_path, 'w') as outfile:
    # Iterate over each line in the input file
    for line in infile:
        # Split the line into columns
        columns = line.split()
        # Select the last two columns
        last_two_columns = columns[-2:]
        # Join the last two columns with a space and write to the output file
        outfile.write(' '.join(last_two_columns) + '\n')

print(f"Processed file has been saved to {output_file_path}")
```
